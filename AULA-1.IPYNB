```python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def falar(self):
        print(f"{self.nome} está falando.")
```

```python
P1 = Pessoa("Ana",22)
P2 = Pessoa("Carlos", 30)

P1.falar()
P2.falar()
```

```python
class Carro:
    def __init__(self, marca, modelo, ano):
        self.marca = marca
        self.modelo = modelo
        self.ano = ano
    
    def ligar(self):
        print(f"{self.marca} {self.modelo} ligado!")

C1 = Carro("Fiat", "Uno", 2015)
C1.ligar()
```

```python
a = 10
b = 3

# operadores adicionais
a // b  # divisão inteira
a % b   # módulo (resto da divisão)
a ** b  # exponenciação (a elevado a b)
```

```python
a = True
b = False

#operadores lógicos
a and b  # False
a or b   # True
not a     # False
```
